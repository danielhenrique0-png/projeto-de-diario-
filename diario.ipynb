```python
# @title
class Cadastro:
  def __init__(self, nome):
    self.nome = nome

  def exibir(self):
    print("Classe pai")


class Usuario(Cadastro):
  def __init__(self):
    super().__init__("Usuários")
    self.usuarios = []
    self.idades = []

  def cadastrar(self, nome, idade):
    self.usuarios.append(nome)
    self.idades.append(idade)

  def exibir(self):
    print("====== Usuários ======")
    for i in range(len(self.usuarios)):
      print(f"Nome: {self.usuarios[i]}, Idade: {self.idades[i]}")
```

```python
# @title
class Anotacao(Cadastro):
  def __init__(self, titulo, texto, data):
    super().__init__(titulo)
    self.texto = texto
    self.data = data

  def exibir(self):
    print(f"Título: {self.nome}")
    print(f"Data: {self.data}")
    print(f"Texto: {self.texto}")
```

```python
# @title
class Diario(Cadastro):
  def __init__(self):
    super().__init__("Diário")
    self.anotacoes = []

  def criar_anotacao(self, titulo, texto, data):
    nova = Anotacao(titulo, texto, data)
    self.anotacoes.append(nova)

  def exibir(self):
    print("====== Anotações ======")
    for i in range(len(self.anotacoes)):
      print(f"\n--- Anotação {i} ---")
      self.anotacoes[i].exibir()

  def excluir_anotacao(self, posicao):
    if posicao in range(len(self.anotacoes)):
      self.anotacoes.pop(posicao)
      print("Excluída!")
    else:
      print("Não encontrada.")
```

```python
# @title
sistema_usuario = Usuario()
diario = Diario()
```
